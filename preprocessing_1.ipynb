```python
def clean_for_labeling(text):
    text = str(text).lower()
    text = html.unescape(text) # decode HTML entity (&quot; -> ")
    text = re.sub(r'<.*?>', ' ', text) # hapus tag HTML (<br>, <a>, </a>, dll)
    text = re.sub(r'http\S+|www\S+', '', text) # hapus tag HTML (<br>, <a>, </a>, dll)
    text = re.sub(r'@\w+', '', text)  # hapus mention
    text = re.sub(r'#\w+', '', text)  # hapus hashtag beserta katanya
    text = re.sub(r'[^a-zA-Z\s]', ' ', text) # hapus karakter aneh tapi biarkan spasi
    text = re.sub(r'\s+', ' ', text).strip() # rapikan spasi

    return text
```

# Preprocessing awal data twitter

```python
df_twitter = pd.read_csv('dataset_twitter_gabungan.csv')
df_twitter.head()
```

```python
df_twitter = df_twitter[['full_text']]
df_twitter.head()
```

```python
df_twitter['text'] = df_twitter['full_text'].apply(clean_for_labeling)

df_twitter = df_twitter.drop_duplicates(subset='text')
df_twitter = df_twitter[df_twitter['text'].str.split().str.len() >= 5]
df_twitter = df_twitter.reset_index(drop=True) #reset index

df_twitter[['full_text', 'text']].head()
```

# Preprocessing awal data youtube

```python
df_youtube = pd.read_csv('dataset_youtube_gabungan.csv')
df_youtube.head()
```

```python
df_youtube = df_youtube[['textDisplay']]
df_youtube = df_youtube.rename(columns={'textDisplay': 'full_text'})
df_youtube.head()
```

```python
# df_youtube['text'] = df_youtube['full_text'].apply(lambda x: clean_for_labeling(x, slang_dict))
df_youtube['text'] = df_youtube['full_text'].apply(clean_for_labeling)

df_youtube = df_youtube.drop_duplicates(subset='text')
df_youtube = df_youtube[df_youtube['text'].str.split().str.len() >= 5]
df_youtube = df_youtube.reset_index(drop=True) #reset index

df_youtube[['full_text', 'text']].head()
```
