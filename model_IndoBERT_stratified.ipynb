##**top 10 bigram**

```python
# Fungsi mengambil top bigram
def get_top_bigrams(text_series, n=10):

    words = ' '.join(
        text_series.astype(str)
    ).split()

    bigram_freq = Counter(
        bigrams(words)
    ).most_common(n)

    bigram_df = pd.DataFrame(
        bigram_freq,
        columns=['bigram', 'frekuensi']
    )

    bigram_df['bigram'] = bigram_df['bigram'].apply(
        lambda x: ' '.join(x)
    )

    return bigram_df

# Ambil bigram tiap stance
pro_bigram = get_top_bigrams(
    df[df['label'] == 'Pro']['text']
)

kontra_bigram = get_top_bigrams(
    df[df['label'] == 'Kontra']['text']
)

netral_bigram = get_top_bigrams(
    df[df['label'] == 'Netral']['text']
)

# Plot
fig, axes = plt.subplots(1, 3, figsize=(22, 8))

# PRO
axes[0].barh(
    pro_bigram['bigram'],
    pro_bigram['frekuensi']
)
axes[0].set_title('Pro')
axes[0].invert_yaxis()

for i, v in enumerate(pro_bigram['frekuensi']):
    axes[0].text(v, i, str(v), va='center')

# KONTRA
axes[1].barh(
    kontra_bigram['bigram'],
    kontra_bigram['frekuensi']
)
axes[1].set_title('Kontra')
axes[1].invert_yaxis()

for i, v in enumerate(kontra_bigram['frekuensi']):
    axes[1].text(v, i, str(v), va='center')

# NETRAL
axes[2].barh(
    netral_bigram['bigram'],
    netral_bigram['frekuensi']
)
axes[2].set_title('Netral')
axes[2].invert_yaxis()

for i, v in enumerate(netral_bigram['frekuensi']):
    axes[2].text(v, i, str(v), va='center')

plt.suptitle(
    'Top 10 Bigram Berdasarkan Stance',
    fontsize=16
)

plt.tight_layout()
plt.show()
```

##Tokenisasi

```python
model_name = "indobenchmark/indobert-base-p1"

tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        padding="max_length",
        truncation=True,
        max_length=128
    )

skf = StratifiedKFold(
n_splits=5,
shuffle=True,
random_state=42
)

fold_results = []
```

#INDOBERT

```python
model = AutoModelForSequenceClassification.from_pretrained(
    model_name,
    num_labels=3
)
model.config.problem_type = "single_label_classification"
```

```python
def compute_metrics(pred):
    labels = pred.label_ids
    preds = pred.predictions.argmax(axis=-1)

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels,
        preds,
        average='weighted'
    )

    acc = accuracy_score(labels, preds)

    return {
        "accuracy": acc,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }
```

#konfigurasi fine tuning

```python
training_args = TrainingArguments(
    output_dir="./results",
    eval_strategy="epoch",
    save_strategy="epoch",
    logging_strategy="epoch",
    num_train_epochs=2,
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    load_best_model_at_end=True
)
```

```python
fold_results = []

best_f1 = 0
best_trainer = None
best_fold = None
best_logs = None

all_y_true = []
all_y_pred = []

for fold, (train_idx, val_idx) in enumerate(
    skf.split(df["text"], df["label"]),
    start=1
):

    print(f"\n========== Fold {fold} ==========")

    train_df = df.iloc[train_idx]
    val_df = df.iloc[val_idx]

    train_dataset = Dataset.from_pandas(
        train_df[["text", "label"]]
    )

    val_dataset = Dataset.from_pandas(
        val_df[["text", "label"]]
    )

    train_dataset = train_dataset.map(
        tokenize_function,
        batched=True
    )

    val_dataset = val_dataset.map(
        tokenize_function,
        batched=True
    )

    train_dataset.set_format(
        type="torch",
        columns=["input_ids", "attention_mask", "label"]
    )

    val_dataset.set_format(
        type="torch",
        columns=["input_ids", "attention_mask", "label"]
    )

    model = AutoModelForSequenceClassification.from_pretrained(
        model_name,
        num_labels=3
    )

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=train_dataset,
        eval_dataset=val_dataset,
        compute_metrics=compute_metrics
    )

    # ======================
    # TRAIN
    # ======================
    trainer.train()

    # ======================
    # PREDIKSI VALIDASI
    # ======================
    pred = trainer.predict(val_dataset)

    y_pred = np.argmax(pred.predictions, axis=1)
    y_true = pred.label_ids

    all_y_true.extend(y_true)
    all_y_pred.extend(y_pred)

    # ======================
    # EVALUASI
    # ======================
    result = trainer.evaluate()

    fold_results.append({
        "fold": fold,
        "accuracy": result["eval_accuracy"],
        "precision": result["eval_precision"],
        "recall": result["eval_recall"],
        "f1": result["eval_f1"]
    })

    print(result)

    # ======================
    # SIMPAN FOLD TERBAIK
    # ======================
    if result["eval_f1"] > best_f1:
        best_f1 = result["eval_f1"]
        best_trainer = trainer
        best_fold = fold

        # simpan log untuk grafik
        best_logs = trainer.state.log_history.copy()
```

```python
results_df = pd.DataFrame(fold_results)

print("\n=== HASIL 5 FOLD ===")
print(results_df)

print("\n=== RATA-RATA ===")
mean_results = results_df.drop(columns=["fold"]).mean()

for metric, value in mean_results.items():
    print(f"{metric}: {value*100:.2f}%")

print(f"\nBest Fold : {best_fold}")
print(f"Best F1   : {best_f1*100:.2f}%")
```

#EVALUATION

```python
print(
    classification_report(
        all_y_true,
        all_y_pred,
        target_names=["Pro", "Kontra", "Netral"],
        digits=4
    )
)

accuracy = accuracy_score(all_y_true, all_y_pred)
precision = precision_score(all_y_true, all_y_pred, average="weighted")
recall = recall_score(all_y_true, all_y_pred, average="weighted")
f1 = f1_score(all_y_true, all_y_pred, average="weighted")

print("\n=== HASIL AKHIR MODEL MENGGUNAKAN STRATIFIED 5-FOLD ===")
print(f"Accuracy  : {accuracy*100:.2f}%")
print(f"Precision : {precision*100:.2f}%")
print(f"Recall    : {recall*100:.2f}%")
print(f"F1-Score  : {f1*100:.2f}%")
```

```python
epochs = []
val_acc = []
val_loss = []
train_loss = []

seen_epochs = set()

for log in best_logs:

    # Train loss
    if "loss" in log:
        train_loss.append(log["loss"])

    # Validation metric
    if "eval_accuracy" in log:
        ep = log["epoch"]

        if ep not in seen_epochs:
            seen_epochs.add(ep)

            epochs.append(ep)
            val_acc.append(log["eval_accuracy"])
            val_loss.append(log["eval_loss"])
```

```python
plt.figure(figsize=(12,5))

# Accuracy
plt.subplot(1, 2, 1)
plt.plot(epochs, val_acc, marker="o", label="Validation Accuracy")
plt.title(f"Accuracy per Epoch (Best Fold = {best_fold})")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.legend()
plt.grid(True)

# Loss
plt.subplot(1, 2, 2)
plt.plot(epochs, val_loss, marker="o", label="Validation Loss")
plt.plot(
    epochs,
    train_loss[:len(epochs)],
    marker="o",
    label="Train Loss"
)

plt.title(f"Loss per Epoch (Best Fold = {best_fold})")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
